## Overview

This notebook extends the Automatic Number Plate Recognition (ANPR) system developed in the previous notebook by evaluating whether the detected Malaysian vehicle registration plate complies with selected regulations issued by the Road Transport Department Malaysia (JPJ).

The notebook first performs license plate recognition using the previously developed ANPR pipeline, which consists of YOLO for license plate detection and CRNN for optical character recognition (OCR). The recognized license plate is then subjected to a series of compliance verification procedures.

The implemented compliance checks include:

- Plate Format Validation
- Character Validation
- Plate Color Validation
- Font Validation
- Character Dimension Measurement
- Character Spacing Measurement
- Plate Type Classification

Finally, the system generates a comprehensive JPJ Compliance Report indicating whether the detected license plate satisfies the selected JPJ requirements.
